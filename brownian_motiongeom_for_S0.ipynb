```python
#running simulation environment of the price with brownian motion
import numpy as np

def simulate_gbm(S0,mu, sigma, T, n_steps, n_paths, seed = None):
    dt = T/n_steps
    rng = np.random.default_rng(seed)
    Z = rng.standard_normal((n_paths, n_steps))
    X = (mu - (sigma**2)/2)*dt + sigma*np.sqrt(dt)*Z
    increments_sum = np.cumsum(X,axis = 1)
    zeros = np.zeros((n_paths,1))
    zeros_increments = np.concatenate([zeros, increments_sum], axis = 1)
    log_Sk = np.log(S0) + zeros_increments
    return np.exp(log_Sk)
     
    
    
```

```python
#testing a function
S = simulate_gbm(S0 = 100, mu = 0.0, sigma = 0.2, T = 30/365, n_steps = 30, n_paths= 100_000, seed = 42 )
print(S.shape)
print(S[:3, :5])
```

```python
S_T = S[:, -1]
print(S_T.mean()) # actual
print(100*np.exp(0.0 * 30/365)) # emperica
```

```python
print(np.log(S_T).var()) # emperical
print(0.2**2 * 30/365) # theoretical
```
